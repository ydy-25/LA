```python
import matplotlib.pyplot as plt
import numpy as np
```

#1)함수 정의와 시각화

```python
def f(x):
  return 2*x+1

x=np.linspace(-10,10,100)
y=f(x)

plt.plot(x,y,label='f(x)=2x+1')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Example of function graph')
plt.legend()
plt.grid(True)
plt.show()
```

#2)단사 함수 여부 확인


```python
def is_injective(f,domain):
  image=set()
  for x in domain:
    fx=f(x)
    if fx in image:
      return False
    image.add(fx)
  return True

domain=range(-10,11)
print("Is it injective func?",is_injective(f,domain))
```
